# MaxQuant phospho to specset, then to library

```python
# python -m pip install "peptdeep[stable]"
# python -m pip install "alphabase[stable,full-stable]"
# python -m pip install -e ~/aiproteomics/dbl_alphapepdeep

from specset.utils import maxquant_to_specset, specset_to_speclib

maxquant_txt_dir = '/d0/tp/specset_experiments/dda_phospho/txt'
output_specset = '/d0/tp/specset_experiments/dda_phospho/txt.specset'
output_speclib = '/d0/tp/specset_experiments/dda_phospho/txt.specset.lib.tsv'

maxquant_to_specset(maxquant_txt_dir, output_specset, additional_columns = {"Score" : "score", "Proteins" : "proteins", "Gene Names" : "genes"})

specset_to_speclib(output_specset, output_speclib, additional_columns = {"proteins" : "Protein.Ids", "genes" : "Genes"}, consensus_method="score")
```
