# Creating spectral library from MaxQuant output

## MaxQuant output to specset to spectral library

````python
from specset.utils import maxquant_to_specset, specset_to_speclib

from pathlib import Path

# change path to the data here
p = Path("/d0/tp/scratch/specset_data/phospho_dda")

print('\nCreate specset from MaxQuant output...\n')
maxquant_txt_dir = str(p / 'txt')
output_specset = str(p / 'txt.specset')
output_speclib = str(p / 'txt.specset.lib.tsv')

maxquant_to_specset(maxquant_txt_dir, output_specset, additional_columns = {"Score" : "score", "Proteins" : "proteins", "Gene Names" : "genes"})

print('\nCreate spectral library from specset...\n')
specset_to_speclib(output_specset, output_speclib, additional_columns = {"proteins" : "Protein.Ids", "genes" : "Genes"}, consensus_method="score")
````

## Create phosphosite and phosphopeptide report from DIA-NN output

To analyze DIA-NN output for phosphoproteomics, package msproteomics is required. Install pybind11 and setuptools prior to installing msproteomics.

```
pip install pybind11
pip install setuptools
pip install msproteomics
```

````python
from pathlib import Path

# change path to the data here
p = Path("/d0/tp/scratch/specset_data/phospho_dia")

fasta_file = str(p / 'uniprot_Human.fasta')

working_dir = p / 'phospho_search_specset_lib'

print(f"\n\nProcess: {working_dir}\n")

!diann_parquet_to_tsv -i {working_dir / 'report.parquet'} -o {working_dir / 'report.tsv'}

!read_diann -o {working_dir / 'report_msproteomics.tsv'} -E Intensities -f {fasta_file} {working_dir / 'report.tsv'}

!sitereport {working_dir / 'report_msproteomics.tsv'} -output_site {working_dir / 'phosphosite-report.txt'} -output_peptide {working_dir / 'phosphopeptide-report.txt'}

print("Done.")

working_dir = p / 'phospho_search_fragpipe_lib'

print(f"\n\nProcess: {working_dir}\n")

!diann_parquet_to_tsv -i {working_dir / 'report.parquet'} -o {working_dir / 'report.tsv'}

!read_diann -o {working_dir / 'report_msproteomics.tsv'} -E Intensities -f {fasta_file} {working_dir / 'report.tsv'}

!sitereport {working_dir / 'report_msproteomics.tsv'} -output_site {working_dir / 'phosphosite-report.txt'} -output_peptide {working_dir / 'phosphopeptide-report.txt'}

print("Done.")

working_dir = p / 'phospho_search_diann_lib'

print(f"\n\nProcess: {working_dir}\n")

!diann_parquet_to_tsv -i {working_dir / 'report.parquet'} -o {working_dir / 'report.tsv'}

!read_diann -o {working_dir / 'report_msproteomics.tsv'} -E Intensities -f {fasta_file} {working_dir / 'report.tsv'}

!sitereport {working_dir / 'report_msproteomics.tsv'} -output_site {working_dir / 'phosphosite-report.txt'} -output_peptide {working_dir / 'phosphopeptide-report.txt'}

print("Done.")
````
