## easypqp and DIA-NN libraries to specsets

````python
from specset.utils import easypqp_to_specset, diannlib_to_specset

from pathlib import Path

p = Path("/d0/tp/scratch/specset_data/plasma_libs")

print('\nCreate specset from easypqp output...\n')

easypqp_to_specset(p / 'dda_experimental_lib.tsv',  p / 'dda_experimental_lib.specset')

print('Create specset from DIA-NN library...\n')

diannlib_to_specset(p / 'report-lib.parquet',  p / 'report-lib.specset')


print('\nSummarize specsets...')
from specset.utils import summarize

summarize(p / 'dda_experimental_lib.specset')
summarize(p / 'report-lib.specset')

print('\nSummarize overlapping specsets...')
summarize(p / 'dda_experimental_lib.specset', p / 'report-lib.specset')
````

## Merge specsets and create spectral library

````python
from specset.utils import merge_specsets

from pathlib import Path

p = Path("/d0/tp/scratch/specset_data/plasma_libs")

print('\nMerge specsets...')

import shutil
if (p / 'merged.specset').exists:
    print(f"Output directory exists. Removing {p / 'merged.specset'}...")
    shutil.rmtree(p / 'merged.specset')

merge_specsets(p / 'report-lib.specset', p / 'dda_experimental_lib.specset', p / 'merged.specset', mode = 'ignore')

from specset.utils import summarize

print('\nSummarize overlapping specsets...')

summarize(p / 'merged.specset', p / 'report-lib.specset')
summarize(p / 'merged.specset', p / 'dda_experimental_lib.specset')

from specset.utils import specset_to_speclib

print('\nCreate merged spectral library...')
specset_to_speclib(p / 'merged.specset', p / 'merged-lib.tsv', additional_columns = {"proteins" : "Protein.Ids", "genes" : "Genes"})
````
